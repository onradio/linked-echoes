# BME Linker Demo

```python
from linker import RAGEntityLinker
from entichecks import MUSIC_GENRE_ECK, MUSIC_ARTIST_ECK
```

```python
GEMINI_API_KEY = ""  # Replace with your actual API key
```

```python
linker = RAGEntityLinker(
    generator_model='gemini-2.5-flash',
    generator_api_key=GEMINI_API_KEY,
)
```

### Testing with one example

```python
candidates = linker._search_candidates(
    search_term="Bill Evans",
    top_k=10,
    verbose=True)

print(len(candidates))
```

```python
candidates
```

```python
validated = linker._validate_candidates(
    candidates=candidates,
    entity_type_eck=MUSIC_ARTIST_ECK,
    verbose=True
)
```

```python
validated
```

```python
print(linker.generate_prompt(
    search_term="Bill Evans",
    context_query="Bill Evans' saxophone",
    entity_type="Music Artist",
    candidates=validated)
)
```

```python
selection = linker.link(
    search_term="Bill Evans",
    entity_type="Music Artist",
    top_k=10,
    verbose=False,
    validate=False,
)
```
